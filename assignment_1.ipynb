### Part 1a / 1b — YOUR CODE HERE

Implement `keep_digits` (default `True`) and support both `str` and `list` of strings.

```python
# Part 1
# Modify remove_sp_char for keep_digits and str | list inputs.
def remove_sp_char(text, keep_digits=True):
    pattern = r"[^a-zA-Z\s]" if keep_digits else r"[^a-zA-Z\s\d]"
    text = re.sub(pattern, "", text)
    return text
```

### Part 2 — YOUR CODE HERE

```python
lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
print(f"Stop Words: {stop_words}")
print("---")

for i, sample in enumerate(sample_corpus):
    print(f"[{i}] IN : {sample}")
    
    # same as previous step, leaves in digits
    line = remove_sp_char(sample)
    print(f"[{i}] OUT: {line}")

    # spell check with TextBlob
    blob = TextBlob(line)
    clean_line = blob.correct()
    print(f"[{i}] spellChecked: {clean_line}")

    # tokenize
    tokens = [t.lower() for t in word_tokenize(str(clean_line))]
    print(f"[{i}] tokens: {tokens}")

    # Lemmatize tokens
    lemmatized_tokens = [lemmatizer.lemmatize(token) for token in tokens]
    print(f"[{i}]Lemmatized text: {' '.join(lemmatized_tokens)}")

    # stop word filter
    filtered_tokens = [word for word in lemmatized_tokens if word not in stop_words]
    print(f"[{i}] filtered tokens: {filtered_tokens}")
    print("---")
```
